# Cálculo de momentos de inercia

## Operaciones con matrices: matriz de inercia de un objeto con forma de T
Tenemos cinco cubos macizos iguales, de masa M y lado d. Se sueldan manera que forman un sólido rígido en forma de T

### i) Cálculo de la posición del centro de masas

Elegimos un sistema de coordenadas cartesiano tal que:

- El origen $(0,0)$ esté en el centro del cubo central de la base de la T.
- El eje $x$ sea horizontal.
- El eje $y$ sea vertical.

Cada cubo tiene su centro en posiciones separadas por una distancia $d$. La disposición típica de la T es:

- Tres cubos en la línea horizontal superior:
  - $(-d, 0)$
  - $(0, 0)$
  - $(d, 0)$
- Dos cubos encima del cubo central:
  - $(0, d)$
  - $(0, 2d)$

El centro de masas de un sistema discreto viene dado por:

$
x_{CM} = \frac{\sum m_i x_i}{\sum m_i}, \quad
y_{CM} = \frac{\sum m_i y_i}{\sum m_i}
$

Como todas las masas son iguales ($m_i = M$), se simplifica a:

$
x_{CM} = \frac{1}{5} \sum x_i, \quad
y_{CM} = \frac{1}{5} \sum y_i
$

Cálculo de $x_{CM}$:

$
x_{CM} = \frac{1}{5} \left[(-d) + 0 + d + 0 + 0\right] = \frac{0}{5} = 0
$

Por simetría, el centro de masas está sobre el eje vertical.

$
y_{CM} = \frac{1}{5} \left[0 + 0 + 0 + d + 2d \right] = \frac{3d}{5}
$

El centro de masas del sistema es:

$
(x_{CM}, y_{CM}) = \left(0, \frac{3d}{5}\right)
$

El centro de masas se encuentra sobre el eje vertical de simetría, a una altura $\frac{3d}{5}$ por debajo del centro del cubo central de la base.

```python
import sympy as sp

# Definimos símbolos
d, M = sp.symbols('d M')

# Número de cubos
N = 5

# Coordenadas explícitas
x1, y1 = -d, 0
x2, y2 = 0, 0
x3, y3 = d, 0
x4, y4 = 0, d
x5, y5 = 0, 2*d

# Centro de masas
x_cm = (M*x1 + M*x2 + M*x3 + M*x4 + M*x5) / (N*M)
y_cm = (M*y1 + M*y2 + M*y3 + M*y4 + M*y5) / (N*M)

# Simplificación
x_cm_simplified = sp.simplify(x_cm)
y_cm_simplified = sp.simplify(y_cm)

x_cm_simplified, y_cm_simplified
```

El resultado coincide con nuestro cálculo previo.

Ya sabemos que el centro de masas es:

$
(x_{CM}, y_{CM}) = \left(0, \frac{3d}{5}\right)
$

Posición de los centros respecto al CM: Restamos el centro de masas a cada coordenada.


\begin{aligned}
(-d, 0) &\to (-d, -\tfrac{3d}{5}) \\
(0, 0) &\to (0, -\tfrac{3d}{5}) \\
(d, 0) &\to (d, -\tfrac{3d}{5}) \\
(0, d) &\to (0, \tfrac{2d}{5}) \\
(0, 2d) &\to (0, \tfrac{7d}{5})
\end{aligned}


Tensor de inercia de cada cubo respecto a su centro:

$
I_{cubo}^{(CM)} = \frac{1}{6} M d^2
$

Por tanto:

$
I_{xx}^{(cubo)} = I_{yy}^{(cubo)} = I_{zz}^{(cubo)} = \frac{1}{6} M d^2
$

Para trasladar al CM global, usamos Steiner:

$
I_{xx} = \sum \left( I_{xx}^{(cubo)} + M y_i^2 \right);
$
$
I_{yy} = \sum \left( I_{yy}^{(cubo)} + M x_i^2 \right);
$
$
I_{xy} = -\sum M x_i y_i.
$

Cálculo de $I_{xx}$:

$
I_{xx} = 5 \cdot \frac{1}{6}Md^2 + M \left[
3\left(\frac{3d}{5}\right)^2 + \left(\frac{2d}{5}\right)^2 + \left(\frac{7d}{5}\right)^2
\right]
$

$
= \frac{5}{6}Md^2 + M \cdot \frac{1}{25} \left(27d^2 + 4d^2 + 49d^2 \right)
$

$
= \frac{5}{6}Md^2 + M \cdot \frac{80}{25} d^2
= \frac{5}{6}Md^2 + \frac{16}{5}Md^2
$

$
I_{xx} = \frac{121}{30} M d^2
$

Cálculo de $I_{yy}$:

$
I_{yy} = 5 \cdot \frac{1}{6}Md^2 + M \left[(-d)^2 + 0 + d^2 + 0 + 0\right]
$

$
= \frac{5}{6}Md^2 + 2Md^2
= \frac{17}{6}Md^2
$

Cálculo de $I_{xy}$:

$
I_{xy} = -M \left[
(-d)\left(-\frac{3d}{5}\right) + 0 + d\left(-\frac{3d}{5}\right) + 0 + 0
\right]
$

$
= -M \left[\frac{3d^2}{5} - \frac{3d^2}{5}\right] = 0
$

Tensor de inercia total:

$
\mathbf{I}_{CM} =
\begin{pmatrix}
\frac{121}{30}Md^2 & 0 & 0 \\
0 & \frac{17}{6}Md^2 & 0 \\
0 & 0 & I_{zz}
\end{pmatrix}
$

donde:

$
I_{zz} = I_{xx} + I_{yy} = \frac{121}{30}Md^2 + \frac{17}{6}Md^2 = \frac{206}{30}Md^2 = \frac{103}{15}Md^2
$

Resultado final

$
\mathbf{I}_{CM} =
\begin{pmatrix}
\frac{121}{30}Md^2 & 0 & 0 \\
0 & \frac{17}{6}Md^2 & 0 \\
0 & 0 & \frac{103}{15}Md^2
\end{pmatrix}
$

Hacemos ahora los mismos cálculos usando SymPy.

```python
# Centro de masas
x_cm = 0
y_cm = 3*d/5

# Coordenadas originales de los centros
x1, y1 = -d, 0
x2, y2 = 0, 0
x3, y3 = d, 0
x4, y4 = 0, d
x5, y5 = 0, 2*d

# Coordenadas respecto al CM
x1p, y1p = x1 - x_cm, y1 - y_cm
x2p, y2p = x2 - x_cm, y2 - y_cm
x3p, y3p = x3 - x_cm, y3 - y_cm
x4p, y4p = x4 - x_cm, y4 - y_cm
x5p, y5p = x5 - x_cm, y5 - y_cm

# Momento de inercia de un cubo respecto a su propio CM
I0 = (1/6) * M * d**2

# Componentes del tensor (Teorema de Steiner)
Ixx = (
    5*I0 +
    M*(y1p**2 + y2p**2 + y3p**2 + y4p**2 + y5p**2)
)

Iyy = (
    5*I0 +
    M*(x1p**2 + x2p**2 + x3p**2 + x4p**2 + x5p**2)
)

Ixy = -M*(
    x1p*y1p + x2p*y2p + x3p*y3p + x4p*y4p + x5p*y5p
)

Izz = (
    (I0 + M*(x1p**2 + y1p**2)) +
    (I0 + M*(x2p**2 + y2p**2)) +
    (I0 + M*(x3p**2 + y3p**2)) +
    (I0 + M*(x4p**2 + y4p**2)) +
    (I0 + M*(x5p**2 + y5p**2))
)

# Simplificación
Ixx_s = sp.simplify(Ixx)
Iyy_s = sp.simplify(Iyy)
Ixy_s = sp.simplify(Ixy)
Izz_s = sp.simplify(Izz)

Ixx_s, Iyy_s, Ixy_s, Izz_s
```

Creamos ahora una matriz con los momentos de inercia calculados.

```python
# Tensor de inercia como matriz
I_tensor = sp.Matrix([
    [Ixx_s, 0, 0],
    [0, Iyy_s, 0],
    [0, 0, Izz_s]
])

I_tensor
```

Se puede factorizar el resultado para que quede más limpio:

```python
sp.factor(I_tensor)
```

## iii) Ejes principales de inercia

El tensor de inercia respecto al centro de masas es:

$
\mathbf{I}_{CM} =
\begin{pmatrix}
\frac{121}{30}Md^2 & 0 & 0 \\
0 & \frac{17}{6}Md^2 & 0 \\
0 & 0 & \frac{103}{15}Md^2
\end{pmatrix}
$

Los ejes principales son los autovectores del tensor de inercia, es decir, las direcciones en las que:

$
\mathbf{I} \, \vec{v} = \lambda \vec{v}
$

donde $\lambda$ son los momentos principales de inercia.

El tensor ya es **diagonal**, por lo que:

- Los autovalores son directamente los elementos de la diagonal.
- Los autovectores son los vectores de la base canónica.

Autovalores (momentos principales):

$
\lambda_1 = \frac{121}{30}Md^2, \quad
\lambda_2 = \frac{17}{6}Md^2, \quad
\lambda_3 = \frac{103}{15}Md^2
$

Los ejes principales coinciden con los ejes coordenados:

- Eje $x$:  
  $
  \vec{e}_x = (1,0,0)
  $

- Eje $y$:  
  $
  \vec{e}_y = (0,1,0)
  $

- Eje $z$:  
  $
  \vec{e}_z = (0,0,1)
  $

Para calcularlo en SymPy, se hace de forma trivial, dado que la matriz ya es diagonal. De una forma general, se calcularía así:

```python
# Tensor de inercia
I = sp.Matrix([
    [121*d**2*M/30, 0, 0],
    [0, 17*d**2*M/6, 0],
    [0, 0, 103*d**2*M/15]
])

# Autovalores
eigenvals = I.eigenvals()

# Autovectores
eigenvects = I.eigenvects()

eigenvals, eigenvects
```

## Integrales de volumen: matriz de inercia de una elipsoide

Vamos a aprender a calcular integrales  de volumen con SymPy. El objetivo es poder calcular los elementos del tensor de inercia de un Sólido Rígido.

Como ejemplo, calcularemos el volumen de una elipsoide y sus momentos principales de inercia. Comenzamos importando las librerías y definiendo las variables.

```python
import sympy as smp
```

Definimos las coordenadas. Será mejor trabajar en coordenadas esféricas.

```python
x, y, z = smp.symbols('x y z') # cartesianas
r, phi, theta = smp.symbols('r phi theta') # esféricas
```

## Volumen de una elipsoide

Calculamos ahora el volumen de una elipsoide de ecuación:

$$ \dfrac{x²}{a} + \dfrac{y²}{b} + \dfrac{z²}{c} = 1. $$

```python
a, b, c = smp.symbols('a b c') # semiejes

x = a * r * smp.sin(theta) * smp.cos(phi)
y = b * r * smp.sin(theta) * smp.sin(phi)
z = c * r * smp.cos(theta)
```

Calculamos el Jacobiano de la transformación cartesianas -» esféricas. Para calcular el Jacobiano hay que calcular la matriz en derivadas parciales respecto de ρ , θ , ϕ. El determinante de esta matriz cuadrada tres por tres da como resultado:

```python
X = smp.Matrix([x, y, z])
Y = smp.Matrix([r, phi, theta])

X.jacobian(Y).det()
```

El Jacobiano es el valor absoluto de esta expresión. Simplificando, obtenemos,

```python
J = smp.Abs(smp.simplify(X.jacobian(Y).det()))
J
```

Así que la integral de volumen,

```python
V_elipsoide = smp.integrate(a*b*c*r**2 * smp.sin(theta), (r, 0, 1), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))
print(V_elipsoide)
```

# Matriz de inercia del elipsoide

Calculamos los elementos de la matriz de inercia mediante su definición por integrales. Suponemos una masa M.

```python
M = smp.symbols('M') # Masa
D = M/V_elipsoide # Densidad de masa

Ixx = smp.integrate(D*(y**2 + z**2)*a*b*c*r**2 * smp.sin(theta), (r, 0, 1), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))
Iyy = smp.integrate(D*(x**2 + z**2)*a*b*c*r**2 * smp.sin(theta), (r, 0, 1), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))
Izz = smp.integrate(D*(y**2 + x**2)*a*b*c*r**2 * smp.sin(theta), (r, 0, 1), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))

Ixy = smp.integrate(-D*x*y*a*b*c*r**2 * smp.sin(theta), (r, 0, 1), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))
Ixz = smp.integrate(-D*x*z*a*b*c*r**2 * smp.sin(theta), (r, 0, 1), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))
Iyz = smp.integrate(-D*y*z*a*b*c*r**2 * smp.sin(theta), (r, 0, 1), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))

Iyx = Ixy
Izx = Ixz
Izy = Iyz

```

Creamos la matriz de inercia respeco al CDM

```python
I_cdm = smp.Matrix([[Ixx, Ixy, Ixz], [Iyx, Iyy, Iyz], [Izx, Izy, Izz]])
I_cdm
```

Podemos simplificar la expresión para que quede más legible:

```python
smp.simplify(I_cdm)
```
