Un péndulo simple de masa $m$, que oscila con periodo $T_0$ cuando se cuelga de un soporte rígido, está colgando de un bloque de masa $2m$. Este bloque se mueve en una línea horizontal en el plano del péndulo por acción de un muelle. El bloque oscila con un periodo $T1 = 0.1T_0$, cuando se quita el péndulo.
* a) Calcula el periodo de oscilación de los dos modos normales.
* b) Si la masa del péndulo está oscilando en el modo más lento con amplitud $A$, ¿cuál es la amplitud del movimiento del bloque?
* c) Si el sistema está inicialmente en reposo y se le da un impulso a la masa del péndulo de forma que empieza a moverse con velocidad $v_o$, obtén la dependencia temporal de la posición del bloque en función del tiempo. Datos: $T_o$ = 1 s; $A$ = 100 mm; $v_0$ = 0.5 m/s.



```python
import sympy as smp

wb, w0 = smp.symbols('w_b w_0')
w2 = smp.symbols('w2')
```

```python
K = smp.Matrix([[wb**2 + (1/2)*w0**2, -(w0**2)/smp.sqrt(2)], [-(w0**2)/smp.sqrt(2), w0**2]])
M = smp.Matrix([[1, 0], [0, 1]])
```

```python
K
```

```python
matriz = (K-w2*M)
```

```python
A, w2 = K.diagonalize()
```
