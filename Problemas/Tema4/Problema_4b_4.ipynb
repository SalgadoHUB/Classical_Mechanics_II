Una varilla homogénea de masa m y longitud L está articulada en el centro de otra igual que cuelga del techo. Halla los modos normales de oscilación (haciendo un esquema que indique la forma de cada modo) y las coordenadas normales.

```python
import sympy as smp

the, phi = smp.symbols('theta varphi')
m, g, L = smp.symbols('m g L')
w2 = smp.symbols('w2')
```

```python
K = m*g*L*smp.Matrix([[1, 0], [0, 1/2]])
M = (1/12)*m*L*L*smp.Matrix([[7, 3], [3, 4]])
```

```python
matriz = (K-w2*M)
matriz
```

```python
smp.det(matriz)
```

```python
equation = smp.det(matriz)
smp.solve(equation, w2)
```
