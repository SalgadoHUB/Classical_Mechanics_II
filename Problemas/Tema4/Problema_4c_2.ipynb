Tres péndulos idénticos se acoplan mediante dos muelles idénticos que unen un péndulo con el siguiente.
* Calcula las frecuencias de los modos normales y las amplitudes relativas en cada modo.
* Si el sistema parte del reposo, con el primer péndulo desplazado una distancia a y los otros dos en su posición de equilibrio, calcula las amplitudes máximas con la que llegarán a oscilar los otros dos péndulos.

```python
import sympy as smp

m, k, R, t = smp.symbols('m k R t')
theta1, theta2, theta3 = smp.symbols('theta1 theta2 theta3')
w2 = smp.symbols('w2') # Frecuencia normal al cuadrado
```

```python
theta1_dot = smp.diff(theta1, t)
theta2_dot = smp.diff(theta2, t)
theta3_dot = smp.diff(theta3, t)
```

```python
T = 0.5 * m * R**2 * (theta1_dot**2 + theta2_dot**2 + theta3_dot**2)
V = 0.5 * k * R**2 * ((theta1-theta2)**2 + (theta2-theta3)**2 + (theta3-theta1)**2)
V.expand()
```

```python
M = m * R**2 * smp.Matrix([[1, 0,0], [0, 1, 0], [0, 0, 1]])
K = k * R**2 * smp.Matrix([[2, -1, -1], [-1, 2, -1], [-1, -1, 2]])

matriz = (K-w2*M)
matriz
```

```python
equation = smp.det(matriz)
smp.solve(equation, w2)
```

```python
A, w2 = K.diagonalize()
w2
```

```python
q1, q2, q3 = smp.symbols('q1 q2 q3')
q1_dot = smp.diff(q1, t)
q2_dot = smp.diff(q2, t)
q3_dot = smp.diff(q3, t)

T = 0.5 * (q1_dot**2 + q2_dot**2 + q3_dot**2)

V = V.subs([(theta1, q1/smp.sqrt(m*R**2)), (theta2, q2/smp.sqrt(m*R**2)), (theta3, q3/smp.sqrt(m*R**2))])
V.simplify()
V.expand()
```

```python
M = smp.Matrix([[1, 0, 0], [0, 1, 0], [0, 0, 1]])
K = (k/m) * smp.Matrix([[2, -1, -1], [-1, 2, -1], [-1, -1, 2]])
w2 = smp.symbols('w2') # Frecuencia normal al cuadrado

matriz = (K - w2*M)
matriz
```
