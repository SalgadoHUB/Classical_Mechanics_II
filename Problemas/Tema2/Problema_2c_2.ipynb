# Cálculo de matrices de inercia

Calcula los momentos y productos de inercia de un cubo uniforme de lado $2L$, cuando los ejes X, Y, Z coinciden con las aristas del cubo.

A la vista de la matriz, encuentra ahora las direcciones de los ejes principales y calcula los momentos principales de inercia para ejes que pasan por el origen.

```python
import sympy as smp

x, y, z = smp.symbols('x y z') # cartesianas

M, L = smp.symbols('M L') # Masa y Lado del cubo

V = L**3 # Volumen del cubo
D = M/V # Densidad de masa
```

## Matriz de inercia del cubo

Calculamos los elementos de la matriz de inercia mediante su definición por integrales. Tomamos origen de coordenadas en un vértice, y colocamos nuestros ejes paralelos a las aristas.

```python
Ixx = smp.integrate(D*(y**2 + z**2), (x, 0, L), (y, 0, L), (z, 0, L))
Iyy = smp.integrate(D*(x**2 + z**2), (x, 0, L), (y, 0, L), (z, 0, L))
Izz = smp.integrate(D*(y**2 + x**2), (x, 0, L), (y, 0, L), (z, 0, L))

Ixy = smp.integrate(-D*x*y, (x, 0, L), (y, 0, L), (z, 0, L))
Ixz = smp.integrate(-D*x*z, (x, 0, L), (y, 0, L), (z, 0, L))
Iyz = smp.integrate(-D*y*z, (x, 0, L), (y, 0, L), (z, 0, L))

Iyx = Ixy
Izx = Ixz
Izy = Iyz

```

Creamos la matriz de inercia respeco al vértice

```python
I_v = smp.Matrix([[Ixx, Ixy, Ixz], [Iyx, Iyy, Iyz], [Izx, Izy, Izz]])
I_v
```

## Diagonalización de la matriz de inercia

Para encontrar los ejes principales de inercia, tenemos que diagonalizar la matriz que hemos encontrado.

```python
A, I_vd = I_v.diagonalize()
I_vd
```

Por lo tanto, los momentos principales son:
- $I_1 = (1/6)ML²$
- $I_2 = I_3 = (11/12)ML²$


Dos de los momentos principales son iguales, así que uno de los ejes principales estaré perfectamente determinado, pero los otros dos no.
Los vectores propiosestán en la matriz $A$:

```python
A
```

Vectores propios:

```python
lambda_1 = A.col(0)
lambda_2 = A.col(1)
lambda_3 = A.col(2)

smp.pprint(lambda_1)
smp.pprint(lambda_2)
smp.pprint(lambda_3)
```

Así que:

- $\lambda_1 = (1,1,1)$
- $\lambda_2 = (-1,1,0)$
- $\lambda_3 = (-1,0,1)$



Otra forma de hacerlo con SymPy es usar las funciones 'eigenvals' y 'eigenvects'.

```python
I_v.eigenvals()
```

Nos indica que el primer autovalor tiene duplicidad 1, y el segundo valor propio tiene duplicidad 2.

En cuanto a los autocevtores:

```python
I_v.eigenvects()
```

Con lo que se obtiene los mismos vectores propios que hemos calculado antes.
