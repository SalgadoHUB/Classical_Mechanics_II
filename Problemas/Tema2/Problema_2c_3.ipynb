# Momento de inercia de una esfera hueca

Calcula el tensor de inercia de una esfera hueca de radio externo $a$ y radio interno $b$ en las siguientes situaciones,
- a) Para ejes con origen en su CDM.
- b) Para ejes paralelos al caso anterior, con origen en un punto de la superficie exterior.

Además, para el último caso,
- c) ¿Cuáles son los ejes principales, cuando el origen de coordenadas está en un punto de la superficie exterior?


```python
import sympy as smp

x, y, z = smp.symbols('x y z') # cartesianas
r, phi, theta = smp.symbols('r phi theta') # esféricas

a, b = smp.symbols('a b') # Radios de la esfera.
```

Podemos calcular el volumen de la esfera hueca restando los volúmenes de dos esferas macizas:

```python
V_a = smp.integrate(r ** 2 * smp.sin(theta), (r, 0, a), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))
V_b = smp.integrate(r ** 2 * smp.sin(theta), (r, 0, b), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))

V_ab = V_a - V_b
V_ab = V_ab.simplify()

print('Volumen esfera hueca con radios a y b:')
V_ab
```

## a) Matriz de inercia respecto CDM

Calculamos los elementos de la matriz de inercia mediante su definición por integrales. Calculamos la matriz para cada esfera, y restamos los elementos.

Primero definimos la transformación entre coordenadas esférica sy cartesianas.

```python
x = r * smp.sin(theta) * smp.cos(phi)
y = r * smp.sin(theta) * smp.sin(phi)
z = r * smp.cos(theta)
```

Ahora calculamos la matriz para la esfera hueca. Los límites de integración serán:
- en $r$: entre $b$ y $a$
- En $\varphi$: entre 0 y 2$\pi$
- En $\theta$: entre 0 y $\pi$

Tenemos que calcular también la densidad, usando la masa total y el volumen de la esfera hueca.

```python
#D = smp.symbols('D') # Densidad de masa
M = smp.symbols('M') # Masa total de la esfera hueca
D = M /V_ab

Ixx = smp.integrate(D*(y**2 + z**2)*r**2 * smp.sin(theta), (r, b, a), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))
Iyy = smp.integrate(D*(x**2 + z**2)*r**2 * smp.sin(theta), (r, b, a), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))
Izz = smp.integrate(D*(y**2 + x**2)*r**2 * smp.sin(theta), (r, b, a), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))

Ixy = smp.integrate(-D*x*y*r**2 * smp.sin(theta), (r, b, a), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))
Ixz = smp.integrate(-D*x*z*r**2 * smp.sin(theta), (r, b, a), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))
Iyz = smp.integrate(-D*y*z*r**2 * smp.sin(theta), (r, b, a), (phi, 0, 2*smp.pi), (theta, 0, smp.pi))

Iyx = Ixy
Izx = Ixz
Izy = Iyz

I_cdm = smp.Matrix([[Ixx, Ixy, Ixz], [Iyx, Iyy, Iyz], [Izx, Izy, Izz]])
I_cdm.simplify()
I_cdm
```
